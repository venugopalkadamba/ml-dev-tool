We aim to build the data generator with many more parameters for injecting bias across simulated demographic groups, and also to create varying biases across train, test, and deployment sets. There will be inputs for how many di↵erent demographic groups to include and how they connect, various underlying distributions to draw from to create the input features (normal, binomial, exponential, Poisson, and others.


The user inputs of data generation system should be the 1) bias type, 2) affected features, and 3)intensity. Based on the given requirements, our data generation system can then do the sampling from differetn demographic groups, to implement the desired bias.

```python
'''
Bias types to add: (we can add others in future)
1. under_representation: data which under-represents a given demographic group
2. selective_label: selective label bias
3. differential_informativeness: features with differential informativeness across demographic groups
4. feature_missingness: data with various levels of feature missingness across demographic groups
5. incorrect_labels: varying levels of incorrect labels across demographic groups
'''
bias_types = {
    1: "under_representation",
    2: "selective_label",
    3: "differential_informativeness",
    4: "feature_missingness",
    5: "incorrect_labels"
}
```

```python
def constructBias(bias_types = 0, affected_features = [], intensity = 0.0):
    # The constructBias function simulates introducing bias into a dataset based on specified parameters.
    
    # Parameters:
    # - bias_types: A integer to denote the bias type.
    #   Possible bias types could be 1,2,3,4,5, as per bias_types.
    # - affected_features: A list of strings, each representing a feature in the dataset that will be affected by the bias.
    #   For example, affected_features = ["age", "race", "income"].
    # - intensity: A float value representing the intensity level of the bias, ranging from 0.0 to 1.0.
    #   For example, an intensity of 0.5 might mean that 50% of the affected data will be influenced by the specified bias.

    bias = [bias_types, affected_features, intensity]
    
    return bias
```
